```python
import xarray as xr

path = r"C:\Users\justi\OneDrive\Documents\GEOL_415\HW Assignments\09-16\GEOS-CF_AirQuality_20180101_0030z.nc4"

ds = xr.open_dataset(path, engine="netcdf4")

print(ds)
```

```python
CO = ds.CO
```

```python
CO.plot()
```

```python
#Elementwise vs matrix operations
CO = ds.CO * 1e6 # convert to ppm

#Plotting basics
#coordinate reference system (map projection)
fig = plt.figure(figsize=(9,6))
ax = plt.axes(projection=ccrs.Robinson())
CO.plot(ax=ax, transform=ccrs.PlateCarree(), 
        cbar_kwargs={'shrink': 0.4}) # This line just makes the axis smaller
ax.coastlines()
ax.gridlines()
```

```python
CO = ds.CO*1e6

fig = plt.figure(figsize=(9,6))
ax = plt.axes(projection=ccrs.Robinson())
CO.plot(ax=ax, transform=ccrs.PlateCarree(),
        norm=colors.LogNorm(), # Normalize the colorbar in log scale
        cbar_kwargs={'shrink': 0.4},)
ax.coastlines()
ax.gridlines()
```

```python
CO.sel(lat=[32.2540], lon=[-110.9742], method="nearest")
```

```python
#Elementwise vs matrix operations
O3 = ds.O3 * 1e9 # convert to ppb

#Plotting basics
#coordinate reference system (map projection)
fig = plt.figure(figsize=(9,6))
ax = plt.axes(projection=ccrs.Robinson())
O3.plot(ax=ax, transform=ccrs.PlateCarree(), 
        cbar_kwargs={'shrink': 0.4}) # This line just makes the axis smaller
ax.coastlines()
ax.gridlines()
```

```python
O3.sel(lat=[34.0522], lon=[-118.2437], method="nearest") #Los Angeles
```

```python
O3.sel(lat=[19.6086], lon=[-155.5038], method="nearest") #Hawaii
```

```python
highest_value = np.max(O3)

print('Highest value:', highest_value)
```

```python
lowest_value = O3.min()  # or min(data)
print(lowest_value)
```

```python
O3 = ds.O3*1e9

fig = plt.figure(figsize=(9,6))
ax = plt.axes(projection=ccrs.Robinson())
O3.plot(ax=ax, transform=ccrs.PlateCarree(),
        cbar_kwargs={'shrink': 0.4},)
ax.coastlines()
ax.gridlines()
```

```python
norm=colors.LogNorm()

```

```python
print(O3.min().values)
print(O3.max().values)
print(O3.isnull().sum().values)
```

```python
print(O3)
```

```python
max_O3 = O3.max()

print("Maximum O3:", max_O3.values)
```

```python
max_value = O3.max()

max_location = O3.where(O3 == max_value, drop=True)

print("Maximum O3:", max_value.values)
print("Latitude:", max_location.lat.values)
print("Longitude:", max_location.lon.values)
```

```python
min_value = O3.min()

min_location = O3.where(O3 == min_value, drop=True)

print("Minimum O3:", min_value.values)
print("Latitude:", min_location.lat.values)
print("Longitude:", min_location.lon.values)
```
