```python
fileName = r"C:\Users\justi\OneDrive\Documents\GEOL_415\Assignments\09-23\ManuaLoa_CO2.csv"
pd.read_csv(fileName)
```

```python
df = pd.read_csv(fileName)
df['average']
```

```python
df_CO2 = df['average']

df_CO2 = df[['decimal date', 'average']]
df_CO2.head()
```

```python
ax1 = df_CO2.plot.scatter(
    x='decimal date',
    y='average',
    figsize=(12, 6),
    s=2,
    color="steelblue",
)

ax1.set_title("Graph of CO2 Concentration in Hawaii")
ax1.set_xlabel("Date")
ax1.set_ylabel("Average CO2")

plt.xticks(rotation=45)
plt.grid(True, which='both')
plt.show()
```

```python
ax1 = df_CO2.plot.scatter(
    x='decimal date',
    y='average',
    figsize=(12, 6),
    s=900,
    c='average',
    cmap='rainbow',
    marker='X',
    alpha=0.9
)

ax1.set_title("CO2 Graph for the Former Sandwich Islands")
ax1.set_xlabel("Date (In decimal)")
ax1.set_ylabel("Average CO2")

plt.xticks(rotation=90)
plt.grid(True, which='both')
plt.show()
```
